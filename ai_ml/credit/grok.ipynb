```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Assuming the dataset is provided as 'credit_risk_dataset.csv'. If not, you can generate synthetic data as below.
# For synthetic data generation (uncomment if needed):
# from sklearn.datasets import make_classification
# X, y = make_classification(n_samples=1000, n_features=10, n_informative=8, n_classes=2, random_state=42)
# df = pd.DataFrame(X, columns=[f'feature_{i}' for i in range(10)])
# df['default'] = y
# df.to_csv('credit_risk_dataset.csv', index=False)

# Load the data
df = pd.read_csv('credit_risk_dataset.csv')

# Explore data types, null values, and basic stats
print("Data Types:\n", df.dtypes)
print("\nNull Values:\n", df.isnull().sum())
print("\nDescriptive Statistics:\n", df.describe())

# Understand features (assuming common credit risk features like age, income, loan_amount, credit_score, employment_status, default)
# Replace with actual column names if different
features = df.columns[:-1]  # Assuming last column is target 'default'

# Visualize distributions
for col in features:
    if df[col].dtype in ['int64', 'float64']:
        sns.histplot(df[col], kde=True)
        plt.title(f'Distribution of {col}')
        plt.show()

# Visualize correlations
plt.figure(figsize=(10, 8))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()

# Outliers visualization using boxplots
for col in features:
    if df[col].dtype in ['int64', 'float64']:
        sns.boxplot(x=df[col])
        plt.title(f'Boxplot of {col}')
        plt.show()

# Save clean dataset (initial, before preprocessing)
df.to_csv('clean_credit_risk_dataset.csv', index=False)
```
